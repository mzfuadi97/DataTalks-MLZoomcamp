## Q6. Median value of horsepower

```python
median_before = df['horsepower'].median()
median_before
```

```python
most_frequent = df['horsepower'].mode()[0]
most_frequent
```

```python
hp_filled = df['horsepower'].fillna(most_frequent)
median_after = hp_filled.median()
median_after
```

```python
if median_after > median_before:
    print('Yes, it increased')
elif median_after < median_before:
    print('Yes, it decreased')
else:
    print('No')
```

## Q7. Sum of weights

```python
X = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].iloc[:7].values
X
```

```python
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv.dot(X.T).dot(y)
w
```

```python
w.sum()
```
